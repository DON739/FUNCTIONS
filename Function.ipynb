1. What is the difference between a function and a method in Python?
-In Python, the **difference between a function and a method** comes down to how they are defined and used:



- **Function**:
- A **function** is a block of code that performs a specific task.
- It is defined using the `def` keyword **outside** of any class.
- It can be called **independently** and doesn’t belong to an object.

**Example**:
```python
def greet(name):
    return f"Hello, {name}!"
```



-  **Method**:
- A **method** is a function that is **associated with an object** (usually defined inside a class).
- It takes `self` as its first parameter (for instance methods), referring to the instance of the class.
- It is **called on an object**.

**Example**:
```python
class Greeter:
    def greet(self, name):
        return f"Hello, {name}!"

g = Greeter()
print(g.greet("Alice"))
```

5. What are iterators in Python and how do they differ from iterables?
- In Python, iterators and iterables are related but distinct concepts used in looping and data traversal.



- Iterable
- An iterable is any object that can be looped over (e.g., in a for loop).
- It implements the __iter__() method and returns an iterator
- Common examples: list, tuple, dict, str.

**Example**:
```python
numbers = [1, 2, 3]
for num in numbers:
    print(num)
```


- Iterator
- An iterator is an object with a state which produces the next item in a sequence when next() is called.
- It implements two methods:
  - __iter__() → returns the iterator object itself
  - __next__() → returns the next value or raises StopIteration

**Example**:
```python
numbers = [1, 2, 3]
it = iter(numbers)  
print(next(it))  
print(next(it))  
```

6. Explain the concept of generators in Python and how they are defined.
- In Python, a generator is a special type of iterator that allows you to generate values on the fly, one at a time, using yield instead of return.


- A generator is a function that remembers its state and resumes where it left off each time it's called.
- Instead of computing and storing all values at once (like a list), it yields one value at a time—lazy evaluation.


- How to Define a Generator

- Using a function with `yield`:

```python
def count_up_to(n):
    count = 1
    while count <= n:
        yield count
        count += 1
```

- Each time you call next() on the generator, it runs until the next yield.

- Using generator expressions (like list comprehensions but with `()`):

```python
squares = (x*x for x in range(5))
```

7. What are the advantages of using generators over regular functions?
- Generators offer several **key advantages** over regular functions, especially when working with **large data sets** or **streams of data**. Here's a clear breakdown:


- 1. Memory Efficiency
- Generators **do not store** the entire result in memory.
- They **yield one item at a time**, which is great for **large or infinite sequences**.

**Example**:
```python
def count_up_to(n):
    for i in range(n):
        yield i
```


- 2. Lazy Evaluation
- Values are generated **on-the-fly** and only when needed.
- Ideal for pipelines and processing huge files or data streams.


- 3. Improved Performance
- **Faster startup** time compared to building full collections.
- Execution pauses at `yield`, resumes on `next()`, so only needed work is done.



- 4. Cleaner, Simpler Code
- Replaces complex iterator classes with **simple functions** using `yield`.

- 5. Infinite Sequences
- Can **represent endless data** without crashing or overloading memory.

**Example**:
```python
def infinite_counter():
    i = 0
    while True:
        yield i
        i += 1
```

9. Explain the purpose and usage of the `map()` function in Python
- The **`map()` function** in Python is used to **apply a function to every item** in an iterable (like a list, tuple, etc.) and return a **new iterator** with the results.




- To **transform** data by applying a **function to each element** in a sequence, without writing a loop.


- Syntax:
```python
map(function, iterable)
```

- `function`: A function that takes one or more arguments.
- `iterable`: One or more iterables (e.g., list, tuple).

```python
nums = [1, 2, 3, 4]
squared = map(lambda x: x ** 2, nums)
print(list(squared))  # [1, 4, 9, 16]
```

- Multiple Iterables:
```python
a = [1, 2, 3]
b = [10, 20, 30]
result = map(lambda x, y: x + y, a, b)
print(list(result))  # [11, 22, 33]
```

10. What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?
- The functions **`map()`**, **`reduce()`**, and **`filter()`** are all built-in functions in Python that allow you to process iterables in different ways. Here's a breakdown of their differences:

- **`map()`**:
- **Purpose**: Applies a function to every item in an iterable and returns an **iterator** with the results.
- **Returns**: An iterator of transformed values.
- **Use case**: When you want to **transform** each item in an iterable (e.g., applying a function to each element).


```python
nums = [1, 2, 3, 4]
squared = map(lambda x: x ** 2, nums)
print(list(squared))  # Output: [1, 4, 9, 16]
```
- **`reduce()`**:
- **Purpose**: Applies a **binary function** cumulatively to the items in an iterable (from left to right) to **reduce** the iterable to a single value.
- **Returns**: A single value (result of cumulative reduction).
- **Use case**: When you need to **combine** items in an iterable into a **single value**, such as computing a sum, product, or maximum.


```python
from functools import reduce

nums = [1, 2, 3, 4]
product = reduce(lambda x, y: x * y, nums)
print(product)  # Output: 24 (1 * 2 * 3 * 4)
```


- **`filter()`**:
- **Purpose**: Filters elements in an iterable by applying a **function** that returns either `True` or `False`, keeping only the elements for which the function returns `True`.
- **Returns**: An iterator containing only the items for which the function returned `True`.
- **Use case**: When you want to **filter out** elements based on a condition.


```python
nums = [1, 2, 3, 4, 5]
even = filter(lambda x: x % 2 == 0, nums)
print(list(even))  # Output: [2, 4]
```

1. Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.

````python
def sum_even_numbers(numbers):
    return sum(num for num in numbers if num % 2 == 0)
numbers = [1, 2, 3, 4, 5, 6]
result = sum_even_numbers(numbers)
print(result)
````

4. Write a Python function that checks if a given number is prime or not from 1 to 200.

````python
def is_prime(n):
    if n < 2 or n > 200:
        return False
    for i in range(2, int(n ** 0.5) + 1):
        if n % i == 0:
            return False
    return True
numbers_to_check = [1, 2, 3, 4, 17, 18, 200]
prime_numbers = [num for num in numbers_to_check if is_prime(num)]
print(prime_numbers)
````

5. Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

````python
class FibonacciIterator:
    def __init__(self, n):
        self.n = n
        self.a, self.b = 0, 1
        self.count = 0
    def __iter__(self):
        return self
    def __next__(self):
        if self.count < self.n:
            fib = self.a
            self.a, self.b = self.b, self.a + self.b
            self.count += 1
            return fib
        else:
            raise StopIteration
fibonacci = FibonacciIterator(10)
for num in fibonacci:
    print(num)
````

6. Write a generator function in Python that yields the powers of 2 up to a given exponent.

````python
def powers_of_two(exponent):
    for i in range(exponent + 1):
        yield 2 ** i
for power in powers_of_two(5):
    print(power)
````

7. Implement a generator function that reads a file line by line and yields each line as a string.

````python
def read_file_line_by_line(file_path):
    with open(file_path, 'r') as file:
        for line in file:
            yield line.strip()
file_path = '/content/Python.txt'
for line in read_file_line_by_line(file_path):
    print(line)
````

9. Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit

````python
def celsius_to_fahrenheit(celsius):
    return (celsius * 9/5) + 32
celsius_temps = [0, 20, 37, 100, -10]
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))
print(fahrenheit_temps)
````

10. Create a Python program that uses `filter()` to remove all the vowels from a given string.


````python
def remove_vowels(s):
    vowels = "aeiouAEIOU"
    return ''.join(filter(lambda char: char not in vowels, s))
input_string = "Hello, World!"
result = remove_vowels(input_string)
print(result)
````


11. Imagine an accounting routine used in a book shop. It works on a list with sublists, which look like this:




Write a Python program, which returns a list with 2-tuples. Each tuple consists of the order number and the
product of the price per item and the quantity. The product should be increased by 10,- € if the value of the
order is smaller than 100,00 €.

Write a Python program using lambda and map.

````python
book_orders_data = [
    ["34587", "Learning Python, Mark Lutz", "4", "40.95"],
    ["98762", "Programming Python, Mark Lutz", "5", "56.80"],
    ["77226", "Head First Python, Paul Barry", "3", "32.95"],
    ["88112", "Einführung in Python3, Bernd Klein", "3", "24.99"],
]
def process_order(order):
    order_number = order[0]
    quantity = int(order[2])
    price_per_item = float(order[3])
    total_price = quantity * price_per_item
    if total_price < 100:
        total_price += 10
    return (order_number, total_price)
result = list(map(lambda order: process_order(order), book_orders_data))
print(result)
````
